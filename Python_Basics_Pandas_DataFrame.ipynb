Question 1 :  Write a Python program to check whether a number is even or odd using conditional statements.

```python
# Function to check if a number is even or odd
def check_even_odd(number):
    if number % 2 == 0:
        return f"{number} is Even."
    else:
        return f"{number} is Odd."

# Example usage:
num = int(input("Enter a number: "))
print(check_even_odd(num))

```

Question 2: Write a Python program to count how many times a number appears in a list.

Example List

# [2,3,4,2,5,2]

```python
# Given list and target number
my_list = [2, 3, 4, 2, 5, 2]
target_number = 2

# Count the occurrences
count = my_list.count(target_number)

print(f"The number {target_number} appears {count} times in the list.")
```

Question 9 : Create a random array of 10 numbers and:

Find max and min

Sort the array

```python
import numpy as np

# Create a random array of 10 numbers (e.g., integers between 1 and 100)
arr = np.random.randint(1, 101, size=10)

print("Original Array:", arr)
print("Maximum value:", arr.max())
print("Minimum value:", arr.min())

# Sort the array
sorted_arr = np.sort(arr)
print("Sorted Array:", sorted_arr)
```

Question 10 : Replace all even numbers with 0:

arr = np.array([1,2,3,4,5,6])

```python
import numpy as np

# Given array
arr = np.array([1, 2, 3, 4, 5, 6])

# Replace all even numbers with 0 using boolean indexing
arr[arr % 2 == 0] = 0

print(arr)
```
